```python
import re
import matplotlib.pyplot as plt
```

```python
# From chatGPT 

# Function to extract numbers in the format "videos: xxxx"
def extract_numbers_from_file(file_path):
    numbers = []
    with open(file_path, 'r') as file:
        for line in file:
            match = re.search(r'videos:\s*(\d+)', line)
            if match:
                numbers.append(int(match.group(1)))
    return numbers

# File path to the text file
file_path = 'your_text_file.txt'

# Extract numbers
numbers = extract_numbers_from_file(file_path)

# Check if any numbers were found
if numbers:
    # Plot the extracted numbers
    plt.plot(numbers, marker='o', linestyle='-', color='b')
    plt.title('Numbers after "videos:"')
    plt.xlabel('Entry Index')
    plt.ylabel('Videos Count')
    plt.grid(True)
    plt.show()
else:
    print("No numbers found in the file.")
```

### Plot Validation Accuracy across epochs to analyze training speed

```python
def extract_numbers_from_file(file_path, in_string):
    numbers = []
    with open(file_path, 'r') as file:
        for line in file:
            match = re.search(rf'"{in_string}":\s*(\d+\.\d+)', line)
            if match:
                numbers.append(float(match.group(1)))
    return numbers
```

```python
def plot_curves(exp_identifier, y_limit):
    # exp_names = ["sact_"+exp_identifier, "sact_uniSpl_B", "sact_baseline_"+exp_identifier]
    exp_names = ["sact_uniSpl_B_boxPT", "sact_uniSpl_B"]

    root_dir = "/home/aabdujyo/scratch/VideoBoxPretraining/checkpoints/"
    files = [root_dir+x+"/log.txt" for x in exp_names]
    print(files)

    # Extract numbers
    numbers = []

    for f in files:
        numbers.append(extract_numbers_from_file(f, "val_acc1"))

    # Plot the extracted numbers
    for i, data in enumerate(numbers):
        plt.plot(data, linestyle='-')
    plt.legend(["PT->FT", "FT"])
    plt.title('Validation Accuracy '+exp_identifier)
    plt.xlabel('Epoch')
    plt.ylabel('Top-1 Accuracy')
    plt.ylim(0,y_limit)
    plt.grid(True)
    plt.show()
```
