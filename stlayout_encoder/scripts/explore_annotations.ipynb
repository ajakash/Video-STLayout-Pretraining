```python
# Getting sact class id (91 classes)

class_ids = moma.get_cids('sact', 0, 'train') # total 91 classes
sact_class_names = moma.get_cnames(cids_sact=class_ids)

# for fname in fnames:
print(fnames[0])
ann_sact = moma.get_anns_sact(ids_sact=[fnames[0][-10:-5]])[0]
print(ann_sact)
print(ann_sact.cname)
print(sact_class_names.index(ann_sact.cname))
```

```python
import torch
```

```python
# Do for train, val, test, and train-val
# For each min 5 and min 10 boxes per sub-activity

for min_boxes in [5, 10]:
    for split in ['train', 'val', 'test', 'train_val']:
        bbox_set_full, bbox_mask_full, sact_label_indices_full = [], [], []

        if split == 'train_val':
            ids_sact = moma.get_ids_sact(split='train')
            ids_sact.extend(moma.get_ids_sact(split='val'))
        else:
            ids_sact = moma.get_ids_sact(split=split)

        for id_sact in ids_sact:
            if len(bbox_dict[id_sact]) < min_boxes:
                continue

            bbox_set = torch.zeros(1187, 116)
            bbox_mask = torch.ones(1187)
            bbox_mask[:len(bbox_dict[id_sact])] = 0

            # Bounding box info [frame_num, x, y, w, h, object_label], mask, and box label
            for i, item in enumerate(bbox_dict[id_sact]):    
                bbox_set[i, item[0]] = 1
                bbox_set[i, 32] = item[1]
                bbox_set[i, 33] = item[2]
                bbox_set[i, 34] = item[3]
                bbox_set[i, 35] = item[4]
                bbox_set[i, 36 + item[5]] = 1 # box label, 80 classes, 32 + 4
            
            bbox_set_full.append(bbox_set.unsqueeze(0))
            bbox_mask_full.append(bbox_mask.unsqueeze(0))

            # Sub-activity label
            ann_sact = moma.get_anns_sact(ids_sact=[id_sact])[0]
            sact_label_indices_full.append(torch.tensor([sact_class_names.index(ann_sact.cname)]).unsqueeze(0))

        
        # bbox_set_full = torch.cat(bbox_set_full, dim=0)
        # bbox_mask_full = torch.cat(bbox_mask_full, dim=0)
        # sact_label_indices_full = torch.cat(sact_label_indices_full, dim=0)
        # print(bbox_set_full.size(), bbox_mask_full.size(), sact_label_indices_full.size())
        # print(bbox_set_full[:,:,32:36], bbox_mask_full, sact_label_indices_full)

        d = {"bbox_sets": torch.cat(bbox_set_full, dim=0),
         "bbox_masks": torch.cat(bbox_mask_full, dim=0),
         "sact_label_indices": torch.cat(sact_label_indices_full, dim=0)}
        
#         torch.save(d, f'/Users/akashaj/Work/activity_moma/detection_data/min{min_boxes}boxes/BoxEncoderTraining/{split}.pt')
        print(split, min_boxes, len(sact_label_indices_full))
        
```
